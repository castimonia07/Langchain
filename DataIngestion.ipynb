```python
## Text Loader

from langchain_community.document_loaders import TextLoader

loader=TextLoader('speech.txt')
loader
```

```python
# easy way to read a text file and convert it into a document object.
#  The TextLoader class takes the path to the text file as an argument and returns a list of Document objects, which can be used for further processing in LangChain.

text_documents=loader.load()
text_documents
```

```python
## Reading a pdf file

'''The PyPDFLoader class is a document loader that can be used to read PDF files 
and convert them into Document objects. It takes the path to the PDF file as an
argument and returns a list of Document objects, which can be used for further 
processing in LangChain.'''

from langchain_community.document_loaders import PyPDFLoader

loader=PyPDFLoader('attention.pdf')
docs=loader.load()
docs
```

```python
type(docs)  # it will return a list of Document objects, which can be used for further processing in LangChain.
```

```python
## Web based document loader

from langchain_community.document_loaders import WebBaseLoader
import bs4
''' The WebBaseLoader class is a document loader that can be used to
read web pages and convert them into Document objects. 
It takes the URL of the web page as an argument and returns a
list of Document objects, which can be used for further processing 
in LangChain.'''

loader=WebBaseLoader (web_paths=("https://lilianweng.github.io/posts/2023-06-23-agent/",),
                    bs_kwargs=dict(parse_only=bs4.SoupStrainer(
                        class_=("post-title", "post-content","post-header","post-footer")
                    ))
                    )
```

```python
loader.load()  # it will return a list of Document objects, which can be used for further processing in LangChain.
```

```python
## Arxiv
from langchain_community.document_loaders import ArxivLoader
import arxiv
from langchain_core.documents import Document
search = arxiv.Search(
    query="1706.03762",
    max_results=2
)

client = arxiv.Client()

docs = [
    Document(
        page_content=f"{result.title}\n\n{result.summary}",
        metadata={
            "title": result.title,
            "authors": [author.name for author in result.authors],
            "published": str(result.published),
            "entry_id": result.entry_id,
        },
    )
    for result in client.results(search)
]

len(docs)
len(docs)
```
