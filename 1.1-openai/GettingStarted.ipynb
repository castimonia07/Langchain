#### Getting started With Langchain And (Gemini AI)

In this quickstart we'll see how to:

- Get setup with LangChain, LangSmith and LangServe
- Use the most basic and common components of LangChain: prompt templates, models, and output parsers.
- Build a simple application with LangChain
- Trace your application with LangSmith
- Serve your application with LangServe

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GEMINI_API_KEY"]=os.getenv("GEMINI_API_KEY")

# Langsmith Tracking
os.environ["LANGCHAIN_API_KEY"]=os.getenv("LANGCHAIN_API_KEY")
os.environ["LANGCHAIN_TRACING"]="true"
os.environ["LANGCHAIN_PROJECT"]=os.getenv("LANGCHAIN_PROJECT")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash")
print(llm)
```

```python
## Input and get response form the model
result=llm.invoke("What is GenAI?")
```

```python
print(result)
```

```python
## Chatprompt template
'''kaise input output kaise krna hai uska template banane ke liye use hota hai.
'''

from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

prompt=ChatPromptTemplate.from_messages([
    ("system","You are an expert in GenAI and you have to answer the user queries in a very simple way."),
    ("user", "{input}")
])
prompt
```

```python
'''Chain ye krta hai ki input ko prompt me dalta hai aur fir 
llm ko call krta hai.Isse hme ek hi step me input ko prompt me 
dalke llm se response mil jata hai.
'''

chain=prompt | llm 

response=chain.invoke({"input":"Can you tell me about LangSmith?"})
print(response)
```

```python
type(response)  # it give the type of respose that is AIMessage
```

```python
## Stroutput Parser

'''Ye kya krta hai ki output ko string me convert krta hai.
'''

from langchain_core.output_parsers import StrOutputParser
output_parser=StrOutputParser()

chain=prompt | llm | output_parser

response=chain.invoke({"input":"Can you tell me about LangSmith?"})
print(response)
```
